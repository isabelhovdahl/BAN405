## Defining a function

A function definition has a **header** and a **body**.

```
def <name>(<parameters>):
    <statements>
    return <value>
```

The header starts with `def`, gives the function a name, lists its **parameters** in brackets, and ends with a **colon**. The body is **indented** underneath, exactly as it was for `if` and for `for`. The `return` statement says what the function hands back.

````python
def to_fahrenheit(celsius):
    fahrenheit = 9 / 5 * celsius + 32
    return fahrenheit
````

Running that cell produced no output, and that is correct. A definition does not *do* anything - it teaches Python a new word. Nothing happens until we **call** the function.

````python
print(to_fahrenheit(4.2))
print(to_fahrenheit(100))
print(to_fahrenheit(-40))
````

One definition, three uses, one place to fix if the formula is wrong.

A call hands back a value like any other expression, so we can store it, or use it inside something larger.

````python
oslo_f = to_fahrenheit(4.2)

print(f"Oslo is {oslo_f:.1f} F")
print(f"The difference is {to_fahrenheit(7.8) - to_fahrenheit(4.2):.1f} degrees")
````

Two things about the name. Functions follow the same **snake case** convention as variables - lowercase words joined by underscores. And a function name should usually be a **verb or a phrase describing what comes back**, because a function *does* something, whereas a variable *is* something: `to_fahrenheit`, `count_words`, `is_valid`.

> 📝 **Note:** A function must be defined **before** it is called. In a notebook that means the `def` cell has to have been run. If you restart the kernel and jump straight to a call, you get a `NameError` telling you the name is not defined - which is the same message you get for a misspelled variable, and means the same thing.

## `return` is not `print`

This is the single most common misunderstanding about functions, and it is worth being explicit about.

Here is the same conversion written to display its answer instead of returning it.

````python
def show_fahrenheit(celsius):
    fahrenheit = 9 / 5 * celsius + 32
    print(f"{celsius} C is {fahrenheit:.1f} F")


show_fahrenheit(4.2)
````

It looks like it worked. It did not - or rather, it did exactly one thing and gave us nothing.
Watch what happens when we try to use the answer.

````python
# TypeError
average = (show_fahrenheit(4.2) + show_fahrenheit(7.8)) / 2
````

The two temperatures were displayed, and then the program failed, because `show_fahrenheit` has no value to give back. `print` puts characters on the screen; it does not hand anything to the code that called it.

**A function that prints is a dead end. A function that returns can be used again.** So the rule is: compute and return in the function, and display in the code that calls it.

````python
print(f"The average is {(to_fahrenheit(4.2) + to_fahrenheit(7.8)) / 2:.1f} F")
````

### A function with no `return` gives back `None`

We have met `None` before: Python's way of saying *there is no value here*.

````python
result = show_fahrenheit(4.2)

print(result)
print(type(result))
````

So a function without a `return` is not a function that returns nothing - it is a function that returns `None`, which is a value, and which will happily travel through your program until it meets something that cannot cope with it.

### `return` leaves the function immediately

Nothing after a `return` in the same block ever runs.

````python
def double(number):
    return number * 2
    print("This line never runs.")


print(double(21))
````

That is not a limitation - it is how a function with several answers is written. Each branch returns as soon as it knows the answer, and there is no need for an `else`.

````python
def describe(number):
    if number > 0:
        return "positive"
    if number < 0:
        return "negative"
    return "zero"


print(describe(4.2))
print(describe(-1))
print(describe(0))
````

### `return` inside a loop leaves the **function**, not just the loop

We know that `break` leaves the innermost loop and nothing more. `return` is stronger: it abandons the loop *and* the function, in one go. In a search that is usually exactly what you want.

````python
def first_negative(values):
    for value in values:
        if value < 0:
            return value

    return None


print(first_negative([4, 9, -2, 7, -5]))
print(first_negative([4, 9, 7]))
````

Notice the last line of that function. Without it, a list containing no negative number would fall off the end of the function body and return `None` **by accident** rather than on purpose. Both versions behave the same way; only one of them says so.

### Returning a condition

A function that answers a yes-or-no question does not need an `if` at all. A comparison is already `True` or `False`, so it can be returned directly.

````python
def is_valid_scale(scale):
    if scale in ("C", "F"):
        return True
    else:
        return False


def is_valid_scale_short(scale):
    return scale in ("C", "F")


print(is_valid_scale("C"), is_valid_scale_short("C"))
print(is_valid_scale("K"), is_valid_scale_short("K"))
````

Four lines become one, and the short version is easier to read aloud: *return whether the scale is C or F*. Whenever you catch yourself writing `if <condition>: return True else: return False`, return the condition.

## Arguments

The names in the function header are its **parameters**. The values we hand over in the call are its **arguments**. There is more than one way to hand them over.

````python
def convert(temperature, scale):
    if scale == "C":
        return 9 / 5 * temperature + 32

    return 5 / 9 * (temperature - 32)
````

By **position**: the first argument goes to the first parameter, the second to the second. Order is the whole story, and getting it round the wrong way is not an error - it is a wrong answer.

````python
print(convert(100, "C"))
print(convert(100, "F"))
````

By **name**: write `parameter=value` and the order stops mattering. It is longer, and it is worth it whenever a call would otherwise be a row of anonymous values.

````python
print(convert(temperature=100, scale="C"))
print(convert(scale="C", temperature=100))
````

The two styles can be mixed, but positional arguments must come first - Python cannot work out what an unnamed value refers to once the order has been abandoned.

A function must be called with all of its parameters filled in.

````python
# TypeError
convert(100)
````

Read that message: it names the function, says how many arguments are missing, and names them. Error messages in Python are usually this good, and reading them is faster than guessing.

## Default values

A parameter can be given a **default** in the header. That makes it optional: supply it and yours is used, leave it out and the default is used.

````python
def convert(temperature, scale="C"):
    if scale == "C":
        return 9 / 5 * temperature + 32

    return 5 / 9 * (temperature - 32)


print(convert(100))
print(convert(100, "F"))
print(convert(100, scale="F"))
````

Defaults are how a function can be simple in the common case and still flexible in the rare one. `print` works this way: `print(x)` ends the line, and `print(x, end=" ")` does not, because `end` has a default of `"\n"`.

Parameters **with** defaults must come after parameters **without** them. Otherwise there would be calls Python could not interpret, so it refuses the definition outright.

````python
# SyntaxError
def convert(scale="C", temperature):
    return 9 / 5 * temperature + 32
````

## Returning more than one value

Separate several values with commas after `return` and the function hands back all of them.

````python
def summarize(values):
    total = sum(values)
    average = total / len(values)
    return total, average


result = summarize([120, 340, 55, 890])

print(result)
print(type(result))
````

What came back is a **tuple** - the immutable cousin of the list we met when looping over `.items()`. And it unpacks the same way: give as many names on the left as there are values, and Python splits them between the names.

````python
total, average = summarize([120, 340, 55, 890])

print(f"Total: {total}, average: {average:.1f}")
````

## Scope: what a function can see

A variable created inside a function is **local** to it. It comes into existence when the function is called, and disappears when the function returns.

````python
def to_fahrenheit(celsius):
    fahrenheit = 9 / 5 * celsius + 32
    return fahrenheit


print(to_fahrenheit(100))
````

````python
# NameError
print(fahrenheit)
````

That is not an inconvenience - it is the point. Because names inside a function are private to it, two functions can use the same name for different things without colliding, and you can write a function without first checking what every other variable in the program is called.

````python
def area_of_square(side):
    result = side * side
    return result


def area_of_circle(radius):
    result = 3.14159 * radius * radius
    return result


print(area_of_square(3))
print(area_of_circle(3))
````

### Global variables, and why to avoid them

A variable defined outside any function is **global**, and a function can see it without being handed it.

````python
vat_rate = 0.25


def add_vat(price):
    return price * (1 + vat_rate)


print(add_vat(100))
````

That works, and it is a bad habit. The function's header says it needs one thing, and the function actually needs two - so you cannot tell what it depends on by looking at it, you cannot reuse it in another program without carrying `vat_rate` along, and the day somebody changes `vat_rate` somewhere else, this function starts returning different numbers with no error and no warning.

The repair is to pass it in. A default value keeps the common case short.

````python
def add_vat(price, vat_rate=0.25):
    return price * (1 + vat_rate)


print(add_vat(100))
print(add_vat(100, vat_rate=0.15))
````

> 💡 **Tip:** A good function is a black box: everything it needs comes in through the brackets, and everything it produces comes out through `return`. If you can read the header and know what the function depends on, you can reuse it anywhere.

### A function can change your list

There is one way in which a function is *not* sealed off, and it catches everybody once.

````python
def add_ten(number):
    number = number + 10
    return number


score = 5

print(add_ten(score))
print(score)
````

`score` is untouched, exactly as scope promised. Now the same thing with a list.

````python
def add_item(basket):
    basket.append("Oat milk")
    return basket


my_basket = ["Coffee beans"]

print(add_item(my_basket))
print(my_basket)
````

The original list changed. This is the fact about names and objects from the very first week, seen from a new angle: **a parameter is another name for the same object**, not a copy of it. With a number there is nothing to change in place, so `number = number + 10` just points the local name at a new number. With a list, `.append` reaches into the object that both names refer to.

If a function should leave its input alone, it has to work on a copy.

````python
def with_extra_item(basket):
    new_basket = basket[:]
    new_basket.append("Oat milk")
    return new_basket


my_basket = ["Coffee beans"]

print(with_extra_item(my_basket))
print(my_basket)
````

> ⚠️ **Warning:** A function that changes its argument *and* returns it is asking to be misunderstood - the return value makes it look like nothing was touched. Pick one: change the caller's object and return nothing, or leave it alone and return a new one.

## Documenting a function

A string written on the first line of the body is a **docstring**. It is not a comment - Python keeps it, and hands it back when somebody asks what the function does.

````python
def to_fahrenheit(celsius):
    """Return a Celsius temperature converted to Fahrenheit."""
    return 9 / 5 * celsius + 32


help(to_fahrenheit)
````

Every function in Python has one, which is why `help` works on code you did not write.

````python
from random import randint

help(randint)
````

A single line is enough for a short function, and it should say what the function **returns**, not how it works. When there are several parameters and their meaning is not obvious, the longer form earns its keep:

```python
def eff_interest_rate(r, n):
    """
    Return the effective annual interest rate.

    Parameters
    ----------
    r : float
        The nominal annual interest rate, as a decimal.
    n : int
        The number of times per year that interest is compounded.

    Returns
    -------
    float
        The effective annual rate, as a decimal.
    """
    return (1 + r / n) ** n - 1
```

`help(name)`. If hovering over your own function shows nothing useful, that is a message about your
docstring.

## Two kinds of wrong

Everything from here on is about programs that stop.

A **syntax error** means Python could not read your code at all. Nothing runs - not even the correct lines above the mistake - because the whole cell has to be understood before any of it can be executed. The `SyntaxError` a few sections ago was one of these.

An **exception** is different. The program was perfectly readable, it started running, and then it hit something it could not do: a word that is not a number, a division by zero, a position past the end of a list. Everything before that point already happened. Everything after it did not.

````python
# ZeroDivisionError
print("This line runs.")

print(10 / 0)

print("This line does not.")
````

### Reading a traceback

The report Python prints when an exception is not dealt with is a **traceback**, and once you write functions it stops being one line long. Here is a program in three layers.

````python
def run_report(reading):
    return f"{to_fahrenheit(reading):.1f} F"


# TypeError
print(run_report("12.5"))
````

There is a way to read that, and it is not top to bottom.

1. **Start at the bottom.** The last line is the actual complaint: `TypeError: can't multiply sequence by non-int of type 'float'`. That is what went wrong.
2. **Then go up, and find the topmost frame that is your own code.** Above the message is a stack of frames, innermost last. The innermost is inside `to_fahrenheit`, which is where it *broke*. The one above is `run_report`, and above that is the line you actually ran.
3. **The bug is usually not where the crash is.** `to_fahrenheit` is fine. The mistake was passing it `"12.5"` - a string - and the string traveled two levels down before anything objected.

That gap between where a program breaks and where it went wrong is the whole reason tracebacks list several frames rather than one.

### Where exceptions come from

An exception is not something Python does to you. It is something code *raises*, using an ordinary statement.

````python
# ValueError
raise ValueError("that is not a valid temperature")
````

That is exactly what happens inside `float` when you hand it `"forty two"`: it tries, it cannot, and it raises a `ValueError` with a message explaining why. You will not often need to write `raise` yourself in this course, but knowing that is all it is makes the rest of this section much less mysterious.

## The ones you will actually meet

An exception has a **type**, and the type is the useful part - it tells you what kind of thing went wrong, and it is what you will name when you start catching them.

| Type | Raised when |
|---|---|
| `ValueError` | the value is the right kind of thing but the wrong value - `float("abc")` |
| `TypeError` | the value is the wrong kind of thing altogether - `"3" + 3` |
| `ZeroDivisionError` | you divided by zero |
| `IndexError` | a list position that does not exist |
| `KeyError` | a dictionary key that does not exist |
| `NameError` | a name that has not been defined - almost always a typo |

Four of them, live.

````python
# ValueError
float("forty two")
````

````python
# TypeError
"3" + 3
````

````python
# ZeroDivisionError
100 / 0
````

````python
# KeyError
capitals = {"Norway": "Oslo", "Sweden": "Stockholm"}

capitals["Denmark"]
````

`ValueError` and `TypeError` are worth telling apart. `float("abc")` is a `ValueError` because a string is a perfectly reasonable thing to hand to `float` - this particular string just is not a number. `float([1, 2])` is a `TypeError`, because a list is not the sort of thing `float` converts at all.

## `try` and `except`

So far, an exception ends the program. `try` / `except` is how we say *if this goes wrong, here is what to do instead*.

```
try:
    <code that might fail>
except <ExceptionType>:
    <what to do if it does>
```

Python runs the `try` block. If nothing goes wrong, the `except` block is skipped entirely. If the named exception is raised, Python abandons the rest of the `try` block and runs the `except` block instead - and then carries on with the program as if nothing had happened.

````python
text = "forty two"

try:
    number = float(text)
    print(f"The number is {number}")
except ValueError:
    print("That is not a number.")

print("The program carries on.")
````

Change `text` to `"42"` and run it again: the `print` inside the `try` runs, the `except` block is skipped, and the last line is the same either way.

Notice the line that did **not** run when it failed. `number = float(text)` raised, so the `print` underneath it never happened - which is the point. The `try` block is a group of statements that only make sense if the first one worked.

Inside a function, this is how we turn a crash into an answer the caller can deal with.

````python
def to_number(text):
    """Return text as a float, or None if it cannot be read as a number."""
    try:
        return float(text)
    except ValueError:
        return None


print(to_number("42"))
print(to_number("3.14"))
print(to_number("forty two"))
````

## Catch the right thing

`except` can be written with no type at all, and it is almost always a mistake. Here is why.

````python
def to_number(text):
    try:
        return float(txt)
    except:
        return None


print(to_number("42"))
````

`"42"` is a perfectly good number, and the function said it was not.

There is a typo in there: the parameter is `text` and the body uses `txt`. That raises a `NameError` - and a bare `except:` catches **everything**, so the typo was quietly converted into "the input was bad". The function will report that every string you ever pass it is invalid, and it will never once tell you why.

Name the exception you are expecting, and anything you were not expecting is free to interrupt you - which is what you want, because being interrupted is how you find out.

````python
def to_number(text):
    try:
        return float(txt)
    except ValueError:
        return None


# NameError
to_number("42")
````

That is the same bug, and now it is a bug report.

### Seeing the message

Writing `as` and a name gives you the exception object itself, and printing it gives you the message Python would have shown.

````python
for text in ["42", "forty two"]:
    try:
        print(float(text))
    except ValueError as error:
        print(f"Could not convert {text}: {error}")
````

## How much should go inside the `try`?

As little as possible. Everything inside a `try` is protected by the `except` underneath it, and protection is exactly what you do not want for code that has no business failing.

````python
text = "199"
quantity = 3

# Too wide
try:
    price = float(text)
    total = price * quantity
    print(f"Total: {total:.2f}")
    print(f"Average: {total / 0:.2f}")
except:
    print("That is not a valid price.")
````

The price converted perfectly well. The real problem is a division by zero four lines further
down - and the program blamed the input and moved on. That block will report a bad price forever,
whatever you type.

Put only the line that can fail inside the `try`, and do the rest outside it.

````python
text = "199"
quantity = 3

try:
    price = float(text)
except ValueError:
    price = None

if price is None:
    print("That is not a valid price.")
else:
    print(f"Total: {price * quantity:.2f}")
````

## The proper way to check for a number

When we met `.isdigit()`, we said it was the wrong tool for anything that could be negative or have decimals, and that there was a proper one which we would come back to. Here they are side by side.

````python
tests = ["42", "-42", "3.14", " 42 ", "", "1e5", "forty two"]

print(f"{'text':<14}{'.isdigit()':<14}float()")

for text in tests:
    try:
        converted = str(float(text))
    except ValueError:
        converted = "ValueError"

    label = '"' + text + '"'
    print(f"{label:<14}{str(text.isdigit()):<14}{converted}")
````

Every row where the two columns disagree is a row where `.isdigit()` gave the wrong answer about a perfectly good number. `-42` is a number. `3.14` is a number. `1e5` is a number - it is 100 000 written the way scientific software writes it.

The reason is not that `.isdigit()` is badly made. It answers the question it was asked, which is *is every character in this string a digit?* That is simply a different question from *can this be read as a number?*, and the second question has an authoritative answer already: hand it to `float` and see. `float` knows about minus signs, decimal points, surrounding spaces and scientific notation, because knowing about them is its whole job.

**`.isdigit()` asks about the text. `float()` performs the conversion you actually wanted, and tells you if it could not.** Whenever the answer you need is a number, do the second one.

`.isdigit()` is not retired. It is still exactly right for a postal code, a house number or a phone number - things that genuinely are digits and nothing else, and where `-42` should be rejected.

## Asking until you get an answer

Every program you have written so far gave up when the input was wrong, or re-prompted only for choices it could check with a membership test. Now we can re-prompt for a **number**, which was the case we could not do.

The cell below asks you to type something. The prompt appears at the **top of the window**, and the cell keeps running until you answer it. Type something that is not a number first, to watch the loop go round.

````python
def get_number(prompt):
    """Ask for a number, repeating the question until one is given."""
    while True:
        text = input(prompt)

        try:
            return float(text)
        except ValueError:
            print("That is not a number. Please try again.")


temperature = get_number("Enter a temperature in Celsius: ")

print(f"That is {to_fahrenheit(temperature):.1f} F")
````

Three things worth noticing in eight lines.

The `return` is **inside** the `try`, **inside** the `while`. It only runs if the conversion succeeded, and when it does it leaves the loop and the function together - which is exactly why we do not need a flag or a `break`.

The loop cannot end any other way. There is no path out of `while True` except that `return`, so the function is incapable of handing back anything that is not a number. Everything downstream can stop checking.

And the prompt is written once. This is the `while True` shape rather than the test-in-the-header shape, for the same reason as before.

## When should you *not* catch an exception?

Having learned `try` / `except`, the temptation is to put it around everything. Resist it.

A `try` / `except` says **two** things: *I know this can go wrong*, and *I know what to do instead*. The second half is the one that matters, and it is the one people forget.

Here is a function that catches an exception it has no answer for.

````python
def average_price(prices):
    try:
        return sum(prices) / len(prices)
    except:
        return 0


print(average_price([199.00, 249.00, 149.00]))
print(average_price(["199.00", "249.00", "149.00"]))
````

Somebody passed prices that were still strings. `sum` cannot add a string to a number, so it raised a `TypeError` — and the bare `except` swallowed it. The function reported an average price of **zero**: a number that looks like data, goes into a report, and is believed.

Without the `try`, that same call stops the program with a `TypeError` naming the function and the line, and the bug is fixed in a minute.

Catch an exception only if you can answer **both** of these:

- **Which** exception is this? If you cannot name it, you are guessing.
- **What** will I do instead? If the answer is "carry on and hope", there is nothing to do.

If you cannot answer both, let it through. An uncaught exception is not your program failing — it is your program telling you exactly what went wrong and where. Catching it without an answer throws that away and replaces it with a wrong number.

The clear case *for* catching is bad input typed by a person: you know it will happen, you know which exception it produces, and asking again is genuinely better than stopping. That is what `to_number` and `get_number` do. Most other cases are not like that.

## A program built out of functions

Once functions exist, a program stops being a list of statements and becomes a small set of named jobs, plus one function that puts them in order. That last one is conventionally called `main`.

````python
def line_total(quantity, unit_price):
    """Return the total for one receipt line."""
    return quantity * unit_price


def format_line(item, quantity, unit_price):
    """Return one formatted receipt line as a string."""
    return f"{item:<20}{quantity:>3} x {unit_price:>8.2f} = {line_total(quantity, unit_price):>9.2f}"


def main():
    """Print a receipt for the items below."""
    items = ["Espresso machine", "Coffee beans", "Oat milk"]
    quantities = [1, 2, 3]
    unit_prices = [4999.00, 149.90, 24.50]

    subtotal = 0

    for item, quantity, price in zip(items, quantities, unit_prices):
        print(format_line(item, quantity, price))
        subtotal += line_total(quantity, price)

    print("-" * 45)
    print(f"{'TOTAL':<20}{subtotal:>25,.2f}")


main()
````

Nothing new is happening - it is the receipt from the last set of exercises. What changed is that each piece has a name and can be tested on its own. `line_total(3, 24.50)` either gives 73.50 or it does not, and you can find that out without printing a receipt.

Notice also that the data lives **inside** `main` rather than at the top of the file. That is the scope lesson applied: there are now no global variables at all, so no function can accidentally depend on one.

### Applying a function to a whole collection

A function written for one value can be used on a whole list with the comprehension we already have.

````python
temperatures_c = [4.2, 18.5, 11.0, 22.3]

temperatures_f = [to_fahrenheit(t) for t in temperatures_c]

print([round(t, 1) for t in temperatures_f])
````

## How to find a bug

Most of your programming time is spent on programs that already run and do the wrong thing. There is a method, and it is duller and more effective than staring.

**1. Read the traceback, bottom first.** The last line says what went wrong. Then walk up to the topmost frame that is your own code. If there is no traceback at all - the program ran and lied to you - go to step 2.

**2. Print the values, not the code.** Whether or not there was a traceback, put a `print` just above the suspect line showing what the variables actually *contain* - and their `type`, because nine times out of ten the answer is that something is a string when you thought it was a number.

**3. Shrink the input until it still fails.** A loop over 400 rows that crashes is a loop over one row that crashes. Find that row and call the function on it alone.

**4. Check what a function returns before you check what it computes.** A great many "wrong answer" bugs turn out to be a function that returned `None`, because it printed instead of returning, or because one branch had no `return`.
